```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
prop_data = pd.read_csv("HousePricePrediction.csv")
prop_data = prop_data.fillna(prop_data.mean(numeric_only=True))
prop_data
```

```python
# one hot encoding

categorical_cols = ['MSZoning', 'LotConfig', 'BldgType', 'Exterior1st']
prop_data = pd.get_dummies(prop_data, columns=categorical_cols)
prop_data
```

```python
# Column Filteratiom
X = prop_data.drop(columns = ["Id","SalePrice"])
y = prop_data["SalePrice"]
```

```python
# training
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=12)
X_train = X_train.fillna(X_train.mean(numeric_only=True))
y_train = y_train.fillna(y_train.mean(numeric_only=True))
model = LinearRegression()
model
```

```python
model.fit(X_train,y_train)
model
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
r2 = r2_score(y_test, y_pred)
r2
```
